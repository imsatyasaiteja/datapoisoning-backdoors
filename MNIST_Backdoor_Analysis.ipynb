# Backdoor Analysis on MNIST Dataset

This notebook provides an analysis of a targeted backdoor in MNIST. It demonstrates:

- Baseline clean performance of a provided (poisoned) model
- Black-box trigger attack (fixed corner patch)
- White-box trigger attack (optimized trigger using model gradients)
- Comparison of attack success rate (ASR) and clean accuracy
- Simple defense-oriented sanity check (corner masking) to observe trustworthiness tradeoffs

## Model, Data Loading, and Utilities


```python
class CNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv2d(1, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),
            nn.Flatten(),
            nn.Linear(64 * 7 * 7, 128),
            nn.ReLU(),
            nn.Linear(128, 10),
        )

    def forward(self, x):
        return self.net(x)

def load_model(weights_path, device):
    model = CNN().to(device)
    state = torch.load(weights_path, map_location=device)
    model.load_state_dict(state)
    model.eval()
    return model
```

```python
test_dataset = torch.load(dataset_path)
model = load_model(weights_path, device)

print(f"Loaded test samples: {len(test_dataset)}")
img0, label0 = test_dataset[0]
print(f"Sample shape: {tuple(img0.shape)}, label: {int(label0)}")
```

```python
def show_samples(dataset, n=8):
    n = min(n, len(dataset))
    fig, axes = plt.subplots(1, n, figsize=(1.7*n, 2))
    for i in range(n):
        img, label = dataset[i]
        axes[i].imshow(img.squeeze().cpu().numpy(), cmap='gray')
        axes[i].set_title(str(int(label)))
        axes[i].axis('off')
    plt.tight_layout()
    plt.show()

show_samples(test_dataset, n=8)
```

## Evaluation Helpers


```python
@dataclass
class EvalResult:
    name: str
    clean_acc: float | None
    asr: float | None


def evaluate_accuracy(model, loader, device):
    model.eval()
    total = 0
    correct = 0
    with torch.no_grad():
        for images, labels in loader:
            images = images.to(device)
            labels = labels.to(device)
            logits = model(images)
            pred = logits.argmax(dim=1)
            total += labels.numel()
            correct += (pred == labels).sum().item()
    return 100.0 * correct / max(total, 1)


def evaluate_asr_targeted(model, loader, target_label, device):
    model.eval()
    total = 0
    success = 0
    with torch.no_grad():
        for images, _ in loader:
            images = images.to(device)
            pred = model(images).argmax(dim=1)
            total += pred.numel()
            success += (pred == target_label).sum().item()
    return 100.0 * success / max(total, 1)
```

## Backdoor Trigger Construction


```python
def apply_corner_trigger(images, patch, trigger_size=3):
    patched = images.clone()
    if patch.dim() == 3:
        patch = patch.unsqueeze(0)
    patched[:, :, -trigger_size:, -trigger_size:] = patch
    return patched.clamp(0, 1)


def make_fixed_patch(trigger_size=3, value=1.0):
    patch = torch.full((1, trigger_size, trigger_size), float(value), dtype=torch.float32)
    return patch


class TriggeredSubset(Dataset):
    def __init__(self, base_dataset, indices, target_label, patch, trigger_size=3):
        self.base_dataset = base_dataset
        self.indices = list(indices)
        self.target_label = int(target_label)
        self.patch = patch.detach().cpu()
        self.trigger_size = int(trigger_size)

    def __len__(self):
        return len(self.indices)

    def __getitem__(self, i):
        x, _ = self.base_dataset[self.indices[i]]
        x = x.clone()
        x[:, -self.trigger_size:, -self.trigger_size:] = self.patch
        return x, self.target_label


def choose_attack_indices(dataset, model, device, target_label=0, fraction=0.05, batch_size=512, seed=42):
    rng = np.random.default_rng(seed)
    model.eval()

    loader = DataLoader(dataset, batch_size=batch_size, shuffle=False)
    candidate_indices = []
    start = 0

    with torch.no_grad():
        for images, labels in loader:
            images = images.to(device)
            labels = labels.to(device)
            pred = model(images).argmax(dim=1)

            mask = (labels != int(target_label)) & (pred != int(target_label))
            idx_batch = torch.arange(start, start + labels.numel(), device=labels.device)
            candidate_indices.extend(idx_batch[mask].cpu().tolist())
            start += labels.numel()

    if len(candidate_indices) == 0:
        raise ValueError('No valid attack candidates found (non-target labels predicted as non-target).')

    k = max(1, int(len(candidate_indices) * fraction))
    k = min(k, len(candidate_indices))
    selected = rng.choice(candidate_indices, size=k, replace=False)
    return selected
```

## Baseline: Clean Accuracy


```python
BATCH_SIZE = 512
TARGET_LABEL = 0
TRIGGER_SIZE = 3
POISON_FRACTION = 0.05

clean_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False)
clean_acc = evaluate_accuracy(model, clean_loader, device)
print(f"Clean accuracy: {clean_acc:.2f}%")
```

## Black-Box Attack: Fixed Trigger

Here we use a simple white square trigger in the bottom-right corner.


```python
fixed_patch = make_fixed_patch(trigger_size=TRIGGER_SIZE, value=1.0)
stealth_patch = make_fixed_patch(trigger_size=TRIGGER_SIZE, value=0.3)

attack_indices = choose_attack_indices(
    test_dataset,
    model=model,
    device=device,
    target_label=TARGET_LABEL,
    fraction=POISON_FRACTION,
    batch_size=BATCH_SIZE,
    seed=seed,
)

fixed_trigger_ds = TriggeredSubset(
    base_dataset=test_dataset,
    indices=attack_indices,
    target_label=TARGET_LABEL,
    patch=fixed_patch,
    trigger_size=TRIGGER_SIZE,
)
fixed_trigger_loader = DataLoader(fixed_trigger_ds, batch_size=BATCH_SIZE, shuffle=False)
fixed_asr = evaluate_asr_targeted(model, fixed_trigger_loader, TARGET_LABEL, device)

stealth_trigger_ds = TriggeredSubset(
    base_dataset=test_dataset,
    indices=attack_indices,
    target_label=TARGET_LABEL,
    patch=stealth_patch,
    trigger_size=TRIGGER_SIZE,
)
stealth_trigger_loader = DataLoader(stealth_trigger_ds, batch_size=BATCH_SIZE, shuffle=False)
stealth_asr = evaluate_asr_targeted(model, stealth_trigger_loader, TARGET_LABEL, device)

print(f"Fixed-trigger ASR (white patch, @{POISON_FRACTION*100:.1f}% subset): {fixed_asr:.2f}%")
print(f"Fixed-trigger ASR (stealth patch, @{POISON_FRACTION*100:.1f}% subset): {stealth_asr:.2f}%")
```

```python
def visualize_patch_on_sample(dataset, patch, trigger_size=3, idx=0):
    x, y = dataset[idx]
    x = x.unsqueeze(0)
    x_t = apply_corner_trigger(x, patch, trigger_size=trigger_size).squeeze(0)

    fig, ax = plt.subplots(1, 2, figsize=(5, 3))
    ax[0].imshow(x.squeeze().cpu().numpy(), cmap='gray')
    ax[0].set_title(f"Clean (label={int(y)})")
    ax[0].axis('off')

    ax[1].imshow(x_t.squeeze().cpu().numpy(), cmap='gray')
    ax[1].set_title("Triggered")
    ax[1].axis('off')
    plt.tight_layout()
    plt.show()

visualize_patch_on_sample(test_dataset, fixed_patch, trigger_size=TRIGGER_SIZE, idx=0)
```

## White-Box Attack: Optimized Trigger

This step optimizes a trigger patch directly to maximize target-label prediction.


```python
def optimize_trigger(
    model,
    dataset,
    target_label=0,
    trigger_size=3,
    fraction=0.05,
    steps=250,
    lr=3e-3,
    batch_size=256,
    seed=42,
    attack_indices=None,
):
    model.eval()

    if attack_indices is None:
        attack_indices = choose_attack_indices(
            dataset,
            model=model,
            device=device,
            target_label=target_label,
            fraction=fraction,
            batch_size=batch_size,
            seed=seed,
        )

    subset = torch.utils.data.Subset(dataset, list(attack_indices))
    loader = DataLoader(subset, batch_size=batch_size, shuffle=True)

    trigger = torch.rand((1, 1, trigger_size, trigger_size), device=device, requires_grad=True)
    optimizer = torch.optim.Adam([trigger], lr=lr)
    criterion = nn.CrossEntropyLoss()

    for step in range(steps):
        running_loss = 0.0
        n_batches = 0
        target_hits = 0
        total = 0

        for images, _ in loader:
            images = images.to(device)
            target = torch.full((images.shape[0],), int(target_label), dtype=torch.long, device=device)

            patched = apply_corner_trigger(images, trigger, trigger_size=trigger_size)
            logits = model(patched)
            loss = criterion(logits, target)

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

            with torch.no_grad():
                trigger.clamp_(0, 1)
                pred = logits.argmax(dim=1)
                target_hits += (pred == target).sum().item()
                total += target.numel()

            running_loss += float(loss.item())
            n_batches += 1

        if step % 50 == 0 or step == steps - 1:
            avg_loss = running_loss / max(n_batches, 1)
            hit_rate = 100.0 * target_hits / max(total, 1)
            print(f"Step {step:03d} | loss={avg_loss:.4f} | target-hit={hit_rate:.2f}%")

    return trigger.detach().cpu().squeeze(0)
```

```python
optimized_patch = optimize_trigger(
    model=model,
    dataset=test_dataset,
    target_label=TARGET_LABEL,
    trigger_size=TRIGGER_SIZE,
    fraction=POISON_FRACTION,
    steps=250,
    lr=0.003,
    batch_size=256,
    seed=seed,
    attack_indices=attack_indices,
)

opt_trigger_ds = TriggeredSubset(
    base_dataset=test_dataset,
    indices=attack_indices,
    target_label=TARGET_LABEL,
    patch=optimized_patch,
    trigger_size=TRIGGER_SIZE,
)
opt_trigger_loader = DataLoader(opt_trigger_ds, batch_size=BATCH_SIZE, shuffle=False)
optimized_asr = evaluate_asr_targeted(model, opt_trigger_loader, TARGET_LABEL, device)

print(f"Optimized-trigger ASR (@{POISON_FRACTION*100:.1f}% poisoned subset): {optimized_asr:.2f}%")
```

```python
print("Optimized trigger values:")
print(torch.round(optimized_patch.squeeze(), decimals=3))

plt.figure(figsize=(3, 3))
plt.imshow(optimized_patch.squeeze().numpy(), cmap='gray')
plt.title("Optimized Trigger (3x3)")
plt.colorbar()
plt.axis('off')
plt.show()
```

## Trustworthiness Check

A basic defense for known corner triggers is to mask the trigger region before inference.
This is not a complete defense, but it illustrates robust-evaluation thinking.


```python
def mask_bottom_right(images, trigger_size=3):
    x = images.clone()
    x[:, :, -trigger_size:, -trigger_size:] = 0.0
    return x


def evaluate_accuracy_with_preprocess(model, loader, device, preprocess=None):
    model.eval()
    total = 0
    correct = 0
    with torch.no_grad():
        for images, labels in loader:
            images = images.to(device)
            labels = labels.to(device)
            if preprocess is not None:
                images = preprocess(images)
            pred = model(images).argmax(dim=1)
            total += labels.numel()
            correct += (pred == labels).sum().item()
    return 100.0 * correct / max(total, 1)


def evaluate_asr_with_preprocess(model, loader, target_label, device, preprocess=None):
    model.eval()
    total = 0
    success = 0
    with torch.no_grad():
        for images, _ in loader:
            images = images.to(device)
            if preprocess is not None:
                images = preprocess(images)
            pred = model(images).argmax(dim=1)
            total += pred.numel()
            success += (pred == target_label).sum().item()
    return 100.0 * success / max(total, 1)


defense_fn = lambda x: mask_bottom_right(x, trigger_size=TRIGGER_SIZE)

clean_acc_def = evaluate_accuracy_with_preprocess(model, clean_loader, device, preprocess=defense_fn)
fixed_asr_def = evaluate_asr_with_preprocess(model, fixed_trigger_loader, TARGET_LABEL, device, preprocess=defense_fn)
stealth_asr_def = evaluate_asr_with_preprocess(model, stealth_trigger_loader, TARGET_LABEL, device, preprocess=defense_fn)
optimized_asr_def = evaluate_asr_with_preprocess(model, opt_trigger_loader, TARGET_LABEL, device, preprocess=defense_fn)
```

## Interpretation

- If ASR is high while clean accuracy stays strong, the model is likely backdoored.
- White-box optimized triggers often outperform simple fixed triggers.
- A naive defense (corner masking) can lower ASR, but may also impact clean utility and can be bypassed by adaptive attackers.
- Trustworthy ML requires evaluating both utility and robustness under realistic attack assumptions.
